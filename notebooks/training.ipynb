```python
import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset
```

## Load and Visualize the data

```python
processed_data_path = '../data/processed/'

class DataEntry:
    def __init__(self, signals, psd, labels, patient_id):
        self.signals = signals
        self.psd = psd
        self.labels = labels
        self.patient_id = patient_id

def load_data(processed_data_path):
    dataset = []
    for entry in os.listdir(processed_data_path):
        if entry.endswith('.npz'):
            data_entry = load_entry(os.path.join(processed_data_path, entry))
            dataset.append(data_entry)

    return dataset

def load_entry(processed_file):
    data = np.load(processed_file)
    print(list(data.keys()))
    signals = data['signals']
    psd = data['psd']
    labels = data['labels']
    patient_id = data['patient']

    return DataEntry(signals, psd, labels, patient_id)

dataset = load_data(processed_data_path)
```
